## Initial Observations

- The dataset contains 7,043 customers and 21 columns.
- Each row represents one telecom customer.
- The target variable is `Churn`, which indicates whether the customer left the company.
- Around 26.5% of customers churned, while around 73.5% did not churn.
- There are no duplicate rows.
- No standard null values were detected initially.
- `TotalCharges` is stored as a string/object even though it represents a numerical value, so it requires further investigation.
- `customerID` is a unique identifier and will not be used as a predictive feature.
- `SeniorCitizen` is stored numerically as 0/1 but conceptually represents a binary categorical feature.

## Data Cleaning

The initial inspection showed that `TotalCharges` was stored as a string because 11 records contained blank values. All of these customers had a tenure of 0 months, indicating that they were new customers who had not yet accumulated charges. Therefore, these blank values are replaced with 0 and the column is converted to numeric.
